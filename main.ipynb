```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv('credit_card_fraud_10k.csv')
df.head()
```

```python
#Checking for null values in the dataset.
df.isnull().sum()
```

```python
#Dropping the unwanted column ('transaction_id') from the dataset.
df.drop('transaction_id', axis=1, inplace=True)
```

```python
df['merchant_category'].value_counts()
```

```python
# Converting the categorical column 'merchant_category' into numerical values using one-hot encoding.
# Convert text categories into 1s and 0s using dtype=int
df = pd.get_dummies(df, columns=['merchant_category'], drop_first=True, dtype=int)
print(df.head())
```

```python
df.head()
```

```python
df['amount']
```

### This dataset has an amount column with large numbers and a foreign_transaction column with tiny numbers (0 or 1). If we leave it like this, the model will think the amount is more important just because the number is bigger. We use a scaler to normalize the data, which shrinks all the numerical columns down to a similar range.

```python
# Selecting the columns/features which are required for scaling inside an variable called 'columns_to_scale'.
columns_to_scale = ['amount', 'transaction_hour', 'device_trust_score', 
                    'velocity_last_24h', 'cardholder_age']
```

```python
#Assigning the StandardScaler() to a variable called 'scaler'.
scaler = StandardScaler()

# Applying the scaler to the selected columns/features.
df[columns_to_scale] = scaler.fit_transform(df[columns_to_scale])
```

### Now the data is perfectly clean for a machine learning model.

```python
#Extacting the cleaned dataset into a new CSV file called 'cleaned_credit_card_fraud_10k.csv'.
df.to_csv('cleaned_credit_card_fraud_10k.csv', index=False)
```
